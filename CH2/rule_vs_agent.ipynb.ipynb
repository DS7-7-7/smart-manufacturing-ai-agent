# 룰 기반

```python
def rule_based_response(sensor_data):
    line_id = sensor_data["line"]
    temperature = sensor_data["temperature"]
    
    if temperature > 90:
        return f"[경고] {line_id}라인 온도 {temperature}도. 임계값 초과. 즉시 조치 요망."
    elif temperature > 80:
        return f"[주의] {line_id}라인 온도 {temperature}도. 관심 필요."
    else:
        return f"[정상] {line_id}라인 온도 {temperature}도."

sensor_data = {"line": "3번", "temperature": 92}
print(rule_based_response(sensor_data))
```

# AI 에이전트 기반

```python
from google import genai
from google.genai.errors import ServerError
import time

client = genai.Client(
    api_key=""
)

def agent_response(sensor_data, context):

    prompt = f"""
당신은 스마트 팩토리 관리 에이전트입니다.

현재 상황:
- 라인: {sensor_data['line']}
- 온도: {sensor_data['temperature']}도
- 시간: {sensor_data['timestamp']}
- 최근 24시간 온도 이력: {context['history']}
- 계절: {context['season']}

다음을 순서대로 수행하세요:
1. 현재 온도가 위험 수준인지 판단
2. 최근 이력과 계절을 고려하여 가능한 원인 추정
3. 관리자에게 전달할 조치 안내 작성
"""

    for attempt in range(3):
        try:
            response = client.models.generate_content(
                model="gemini-3.8-flash",
                contents=prompt
            )

            return response.text

        except ServerError:
            if attempt < 2:
                time.sleep(2 ** attempt)
            else:
                return "현재 Gemini 서버가 혼잡합니다. 잠시 후 다시 시도해주세요."


sensor_data = {
    "line": "3번",
    "temperature": 92,
    "timestamp": "2025-08-15 14:30"
}

context = {
    "history": "지난 24시간 평균 85도, 오후 2시경 급상승",
    "season": "여름 성수기, 외기 온도 32도"
}

if __name__ == "__main__":
    print(agent_response(sensor_data, context))
```
